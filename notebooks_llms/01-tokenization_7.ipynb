# Claude Tokenizer

```python
tokenizer = GPT2TokenizerFast.from_pretrained("Xenova/claude-tokenizer")
```

```python
tokenizer.vocab_size
```

```python
tokenizer.encode("this is some text")
```

```python
words = ["hypothetical", 
         " hypothetical", 
         "hypothetical ", 
         " hypothetical ",]

for word in words:
    print(f"'{word}' -> {tokenizer.encode(word)} -> {[tokenizer.decode(t) for t in tokenizer.encode(word)]}")

# The tokenizer splits diffferently hypothetical depending on the presence of spaces. 
```
