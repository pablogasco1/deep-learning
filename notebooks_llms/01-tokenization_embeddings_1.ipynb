```python
import re
import numpy as np
from typing import List
import matplotlib.pyplot as plt
```

```python
text = ["All that we are is the result of what we have thought", 
        "To be or not to be that is the question",
        "Be yourself everyone else is already taken",]
```

```python
# separate into words
joined_text = " ".join(text).lower()
all_words = re.split(r'\s+', joined_text)
all_words
```

```python
# create a vocabulary
vocab = set(all_words)
vocab
```

```python
print(f"Number of unique words: {len(vocab)}")
```

```python
# create an encoder and decoder
word2idx = {word: idx for idx, word in enumerate(vocab)} # encoder
idx2word = {idx: word for word, idx in word2idx.items()} # decoder

print(f"Encoder (word to index): {word2idx}")
print(f"Decoder (index to word): {idx2word}")
```

```python
def encoder_function(word : List[str]):
    return [word2idx.get(w, None) for w in word]

def decoder_function(idx : List[int]):
    return [idx2word.get(i, None) for i in idx]

encoder_function(["thought", "so", "to"]), decoder_function([12, 13, 90, 5])
```

```python
 # select random words to create random sentences
rand_idx = np.random.choice(len(vocab), size=5, replace=False)
rand_words = [idx2word[idx] for idx in rand_idx]
print(" ".join(rand_words))
```

```python
# translate the text into numbers
text_as_int = [word2idx[word] for word in all_words]
text_as_int
```

```python
# and numbers back to words
text_as_words = [idx2word[idx] for idx in text_as_int]
text_as_words
```

```python
# what words are in the context of "to"

target_word = "to"
target_idx = word2idx[target_word]

# find the indices of the target word in the text
target_indices = np.where(np.array(text_as_int) == target_idx)[0]
print("Target indices:", target_indices)

# print the context words (1 before and 1 after)
context_size = 1
for idx in target_indices:
    context_words = text_as_words[idx - context_size : idx + context_size + 1]
    print(f"Context for '{target_word}' at index {idx}: {context_words}")
```

```python
len(all_words), len(vocab)

# create a one-hot encoding for all words in the vocabulary without assuming any order
word_matrix = np.zeros((len(all_words), len(vocab)), dtype=int)
for idx, word in enumerate(all_words):
    word_idx = word2idx[word]
    word_matrix[idx, word_idx] = 1

print("One-hot encoded word matrix:")
print(f"Shape of the word matrix: {word_matrix.shape}")
print(word_matrix)
```
